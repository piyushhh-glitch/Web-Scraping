```python
import pandas as pd 
from bs4 import BeautifulSoup
import requests
```

```python
URL="https://books.toscrape.com/"
```

```python
res=requests.get(URL)
```

```python
res
```

```python
with open("scraped_data/books.html","w") as f:
    f.write(res.text)
```

```python
with open("scraped_data/books.html") as f:
    html_content=f.read()
```

```python
# html_content
```

```python
soup=BeautifulSoup(html_content,"lxml")
```

```python
# soup
```

```python
books=soup.select("article.product_pod")[:30]
```

```python
all_books=[]
```

```python
for book in books:
    Title=book.select_one("h3").get_text()
    print(Title)
    Price=book.select_one("p.price_color").get_text(strip=True)
    print(Price)
    Availability=book.select_one("p.instock.availability").get_text(strip=True)
    print(Availability)
    Rating=book.select_one("p.star-rating")["class"][1]
    print(Rating)
    all_books.append((Title,Price,Availability,Rating))
```

```python
all_books
```

```python
df=pd.DataFrame(all_books,columns=["Title","Price","Availability","Rating"])
```
