### Web Scraping for Laptops

```python
import pandas as pd 
import requests 
from bs4 import BeautifulSoup
```

```python
URL="https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops"
```

```python
res=requests.get(URL)
```

```python
res.status_code
```

```python
print(res)
```

```python
with open("scraped_data/laptops.html","w") as f:
    f.write(res.text)
```

```python
with open("scraped_data/laptops.html","r") as f:
    html_content=f.read()
```

```python
soup=BeautifulSoup(html_content,"lxml")
```

```python
soup
```

```python
laptops=soup.select("div.product-wrapper.card-body")[:20]
```

```python
laptops
```

```python
final_list=[]
```

```python
for laptop in laptops:
    Name=laptop.select_one("a.title").get_text(strip=True)
    print(Name)
    Price=laptop.select_one("span[itemprop='price']").get_text()
    print(Price)
    Description=laptop.select_one("p.description.card-text").get_text()
    print(Description)
    Rating=laptop.select_one("p[data-rating]")["data-rating"]
    print(Rating)
    Review=laptop.select_one("p.review-count").get_text().strip()
    print(Review)

    final_list.append((Name,Price,Description,Rating,Review))

```

```python
final_list
```

```python
df=pd.DataFrame(final_list)
```
